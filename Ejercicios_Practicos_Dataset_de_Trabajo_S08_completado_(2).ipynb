## Ejercicio 11: Encontrar valores mayores

Crea `obtener_mayores(valores, limite)` que retorne una nueva lista con los valores mayores que el límite. Pruébala con `[12, 45, 67, 23, 80]` y límite `40`.


```python
def obtener_mayores(valores, limite):
    mayores = []                  # lista nueva, empieza vacía
    for valor in valores:
        if valor > limite:
            mayores.append(valor) # guardamos solo los que superan el límite
    return mayores

# Prueba
print(obtener_mayores([12, 45, 67, 23, 80], 40))   # esperado: [45, 67, 80]
```

## Ejercicio 12: Calcular promedio

Implementa `calcular_promedio(valores)`. Si la lista está vacía, retorna `None`; de lo contrario, retorna el promedio.


```python
def calcular_promedio(valores):
    # Si la lista está vacía no se puede dividir: retornamos None
    if len(valores) == 0:
        return None

    suma = 0
    for valor in valores:
        suma = suma + valor
    return suma / len(valores)

# Pruebas
print(calcular_promedio([10, 20, 30]))   # esperado: 20.0
print(calcular_promedio([]))             # esperado: None
```

## Ejercicio 19: Resumen de un conjunto

Implementa `generar_resumen(valores)` que retorne cantidad, suma, promedio, mínimo y máximo. Decide qué retornar para una lista vacía.


```python
def generar_resumen(valores):
    # Lista vacía: no hay nada que resumir, retornamos None
    if len(valores) == 0:
        return None

    cantidad = len(valores)
    suma = 0
    minimo = valores[0]   # partimos del primer valor
    maximo = valores[0]

    for valor in valores:
        suma = suma + valor
        if valor < minimo:
            minimo = valor
        if valor > maximo:
            maximo = valor

    promedio = suma / cantidad

    return {
        "cantidad": cantidad,
        "suma": suma,
        "promedio": promedio,
        "minimo": minimo,
        "maximo": maximo,
    }

# Pruebas
print(generar_resumen([22, 45, 70, 38, 85]))
print(generar_resumen([]))   # None
```

## Ejercicio 20: Aplicación al Dataset de Trabajo

Selecciona una columna numérica de tu Dataset de Trabajo y crea una lista con una muestra de al menos 10 valores. Implementa funciones para calcular suma, promedio, cantidad de valores por categoría o condición, y una bandera para detectar si algún valor cumple una regla. Explica cómo adaptaste el ejercicio a tus datos.


```python
# ---------- 1. Datos reales: población 2025 por localidad ----------
# Paso a paso:
datos_2025 = datos[datos["ANO"] == 2025]                              # solo el año 2025
localidades = datos_2025[datos_2025["CODIGO_LOCALIDAD"] != 0]         # quitamos "Bogotá" (es el total)
poblacion_por_localidad = localidades.groupby("NOMBRE_LOCALIDAD", sort=False)["POBLACION"].sum()

# Pasamos los resultados a listas normales de Python
nombres = poblacion_por_localidad.index.tolist()   # nombres de las localidades
valores = poblacion_por_localidad.tolist()         # población de cada una

print("Cantidad de localidades:", len(valores))
for i in range(len(valores)):
    print(i, nombres[i], "->", valores[i])


# ---------- 2. Funciones ----------
def calcular_suma(valores):
    suma = 0
    for valor in valores:
        suma = suma + valor
    return suma


def calcular_promedio(valores):
    if len(valores) == 0:
        return None
    return calcular_suma(valores) / len(valores)


def clasificar_localidad(poblacion):
    # Regla adaptada a población (habitantes)
    if poblacion >= 500000:
        return "GRANDE"
    elif poblacion >= 150000:
        return "MEDIANA"
    else:
        return "PEQUEÑA"


def contar_por_categoria(valores):
    grandes = 0
    medianas = 0
    pequenas = 0
    for valor in valores:
        categoria = clasificar_localidad(valor)
        if categoria == "GRANDE":
            grandes = grandes + 1
        elif categoria == "MEDIANA":
            medianas = medianas + 1
        else:
            pequenas = pequenas + 1
    return grandes, medianas, pequenas


def existe_valor_mayor(valores, limite):
    # Bandera: True si al menos un valor supera el límite
    encontrado = False
    for valor in valores:
        if valor > limite:
            encontrado = True
    return encontrado


# ---------- 3. Resumen ----------
suma = calcular_suma(valores)
promedio = calcular_promedio(valores)
grandes, medianas, pequenas = contar_por_categoria(valores)
hay_millon = existe_valor_mayor(valores, 1000000)

print("\nRESUMEN 2025")
print("Población total (suma):", f"{suma:,}")
print("Promedio por localidad:", f"{promedio:,.0f}")
print("Localidades GRANDES (>= 500.000):  ", grandes)
print("Localidades MEDIANAS (150.000 a 499.999):", medianas)
print("Localidades PEQUEÑAS (< 150.000):  ", pequenas)
print("¿Alguna localidad supera 1.000.000 de habitantes?", hay_millon)

# ---------- 4. Verificación contra el total oficial de Bogotá ----------
total_bogota = datos[(datos["ANO"] == 2025) & (datos["CODIGO_LOCALIDAD"] == 0)]["POBLACION"].sum()
print("\nTotal de Bogotá en el dataset:", f"{total_bogota:,}")
print("¿Coincide con nuestra suma?", suma == total_bogota)
```

3. **Suma y promedio:** reutilicé `calcular_suma()` y `calcular_promedio()` de los ejercicios anteriores, sin cambios.
4. **Categorías:** cambié las reglas ALTO/NORMAL/BAJO (pensadas para valores entre 0 y 100) por reglas de población: **GRANDE** (500.000 o más), **MEDIANA** (150.000 a 499.999) y **PEQUEÑA** (menos de 150.000). El conteo usa un contador por categoría.
