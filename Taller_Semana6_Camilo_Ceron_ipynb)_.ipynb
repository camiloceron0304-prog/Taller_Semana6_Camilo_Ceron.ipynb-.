## 2. Reto 2 — Contador por categoría

### 2.1 Datos de práctica
Simula el conteo de tres categorías exhaustivas y mutuamente excluyentes (A, B y C).

````python
# Muestra de prueba con categorías simuladas
categorias_practica = ['A', 'B', 'A', 'C', 'B', 'A', 'A']

# Inicializar contadores
cont_a = 0
cont_b = 0
cont_c = 0

for cat in categorias_practica:
    if cat == 'A':
        cont_a += 1
    elif cat == 'B':
        cont_b += 1
    else:
        cont_c += 1

# Validación de consistencia: A + B + C = TOTAL
total_calculado = cont_a + cont_b + cont_c
try:
    assert total_calculado == len(categorias_practica), "Error: La suma de categorías no coincide con el total."
    print("===== CONTEO POR CATEGORÍA COMPLETADO =====")
    print(f"Categoría A: {cont_a}")
    print(f"Categoría B: {cont_b}")
    print(f"Categoría C: {cont_c}")
    print(f"Validación (A+B+C = Total): {total_calculado} == {len(categorias_practica)}")
except AssertionError as e:
    print(e)
````

### 2.2 Adaptación al Dataset Individual

````python
# Estructura: [ANO, CODIGO_UPL, NOMBRE_UPL, SEXO, EDAD, ORDEN_MCV, CURSODEVIDA, ORDEN_GRUPO_EDAD, GRUPOEDAD, POBLACION]
datos_ninguno = [   # Ningún registro supera 5000
    ['2035', '8', 'Britalia', 'Hombres', '0', '1', 'Primera Infancia', '1', '0 a 4', '724'],
    ['2035', '8', 'Britalia', 'Mujeres', '0', '1', 'Primera Infancia', '1', '0 a 4', '646'],
    ['2035', '8', 'Britalia', 'Hombres', '85', '6', 'Vejez', '18', '85 o más', '2558'],
]

datos_un_solo = [   # Solo el último (6077) supera 5000
    ['2035', '8', 'Britalia', 'Hombres', '0', '1', 'Primera Infancia', '1', '0 a 4', '724'],
    ['2035', '8', 'Britalia', 'Mujeres', '0', '1', 'Primera Infancia', '1', '0 a 4', '646'],
    ['2035', '8', 'Britalia', 'Mujeres', '85', '6', 'Vejez', '18', '85 o más', '6077'],
]

def existe_poblacion_alta(registros):
    bandera = False
    for fila in registros:
        if int(fila[INDICE_POBLACION]) > UMBRAL_ALERTA:
            bandera = True
    return bandera

print("Caso ninguno cumple  ->", existe_poblacion_alta(datos_ninguno))
print("Caso un solo cumple  ->", existe_poblacion_alta(datos_un_solo))

assert existe_poblacion_alta(datos_ninguno) == False
assert existe_poblacion_alta(datos_un_solo) == True
print("Pruebas de frontera superadas")
````

## 7. Retos 10 y 11 — Integración y Resumen Final del Dataset

Un único ciclo que recorre `mi_dataset_lista` y reutiliza la función `clasificar_registro` (sin duplicar código). Se empaqueta en una función para poder reutilizarla también en la tabla de casos de prueba.

````python
def clasificar_registro(fila):
    # Clasifica un registro según su población (función reutilizable)
    poblacion = int(fila[INDICE_POBLACION])
    if poblacion > UMBRAL_ALERTA:
        return "Alerta"      # Más de 5000 habitantes
    elif poblacion >= 100:
        return "Media"       # Entre 100 y 5000
    else:
        return "Baja"        # Menos de 100


def resumir_dataset(registros, anio=None):
    # Inicialización global: contadores, acumulador y bandera
    total = 0
    hombres = 0
    mujeres = 0
    alerta = 0
    media = 0
    baja = 0
    suma_poblacion = 0
    hay_alerta = False

    # Ciclo principal de procesamiento
    for fila in registros:
        total += 1

        if fila[INDICE_SEXO] == "Hombres":
            hombres += 1
        else:
            mujeres += 1

        nivel = clasificar_registro(fila)
        if nivel == "Alerta":
            alerta += 1
            hay_alerta = True
        elif nivel == "Media":
            media += 1
        else:
            baja += 1

        # anio=None suma todos los años; si se indica, solo ese año
        if anio is None or fila[INDICE_ANO] == str(anio):
            suma_poblacion += int(fila[INDICE_POBLACION])

    return {"total": total, "hombres": hombres, "mujeres": mujeres,
            "alerta": alerta, "media": media, "baja": baja,
            "suma": suma_poblacion, "bandera": hay_alerta}


resumen = resumir_dataset(mi_dataset_lista, anio=2025)

print("===== RESUMEN DEL DATASET REAL =====")
print(f"Total de registros: {resumen['total']:,}")
print(f"Hombres: {resumen['hombres']:,}  |  Mujeres: {resumen['mujeres']:,}")
print(f"Nivel Alerta (>5000): {resumen['alerta']}  |  Media: {resumen['media']:,}  |  Baja (<100): {resumen['baja']:,}")
print(f"Población total de Bogotá en 2025: {resumen['suma']:,}")
print(f"¿Existe algún registro en alerta?: {resumen['bandera']}")

# Validaciones de consistencia
assert resumen['hombres'] + resumen['mujeres'] == resumen['total']
assert resumen['alerta'] + resumen['media'] + resumen['baja'] == resumen['total']
print("Validaciones de consistencia superadas")
````
